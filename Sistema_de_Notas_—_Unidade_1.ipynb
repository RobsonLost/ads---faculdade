```python
# Função para cadastrar as notas do aluno
def cadastrar_notas():
    """
    Solicita as notas ao usuário e armazena em uma lista.
    A digitação termina quando o usuário pressiona Enter sem digitar nada.
    """
    notas = []  # Lista que vai guardar as notas
    print("=== Cadastro de Notas ===")
    print("Digite as notas do aluno (ou pressione Enter sem digitar nada para finalizar):")

    while True:
        entrada = input(f"Digite a {len(notas) + 1}ª nota: ")

        # Se o usuário apertar Enter sem digitar, encerra o cadastro
        if entrada.strip() == "":
            if len(notas) == 0:
                print("⚠️  Nenhuma nota foi cadastrada. Digite pelo menos uma nota.")
                continue
            else:
                break

        # Tenta converter a entrada para número
        try:
            nota = float(entrada)
            if nota < 0 or nota > 10:
                print("⚠️  Nota inválida! Digite um valor entre 0 e 10.")
            else:
                notas.append(nota)
                print(f"✅ Nota {nota:.1f} adicionada com sucesso!")
        except ValueError:
            print("⚠️  Entrada inválida! Digite apenas números.")

    return notas


# Função para calcular a média das notas
def calcular_media(notas):
    """
    Recebe a lista de notas e retorna a média aritmética.
    """
    if len(notas) == 0:
        return 0
    return sum(notas) / len(notas)


# Função para determinar a situação do aluno
def determinar_situacao(media):
    """
    Retorna 'Aprovado' se a média for maior ou igual a 7.
    Caso contrário, retorna 'Reprovado'.
    """
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"


# Função para exibir o relatório final
def exibir_relatorio(notas, media, situacao):
    """
    Mostra na tela as notas, a média e a situação do aluno.
    """
    print("\n" + "=" * 45)
    print("        RELATÓRIO FINAL DO ALUNO")
    print("=" * 45)
    print(f"Notas cadastradas : {notas}")
    print(f"Quantidade de notas: {len(notas)}")
    print(f"Média final        : {media:.2f}")
    print(f"Situação           : {situacao}")
    print("=" * 45)

    if situacao == "Aprovado":
        print("🎉 Parabéns! O aluno foi APROVADO!")
    else:
        print("📚 O aluno foi REPROVADO. É preciso estudar mais!")
    print("=" * 45)


# Função principal que organiza a execução do programa
def main():
    """
    Função principal: chama as outras funções na ordem correta.
    """
    print("\n" + "=" * 45)
    print("   SISTEMA DE GESTÃO DE NOTAS - PYTHON")
    print("=" * 45)

    # Passo 1: Cadastrar as notas
    notas = cadastrar_notas()

    # Passo 2: Calcular a média
    media = calcular_media(notas)

    # Passo 3: Determinar a situação
    situacao = determinar_situacao(media)

    # Passo 4: Exibir o relatório final
    exibir_relatorio(notas, media, situacao)


# Executa o programa principal
if __name__ == "__main__":
    main()
```
